```python
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline

from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
```

```python
data = pd.read_csv('CH.SC.U4CSE24156_EX-11.4_1(PCA).csv')

print(data.shape)
print(data.head())
```

```python
data.info()
```

```python
X = data[
    [
        'LotArea',
        'OverallQual',
        'YearBuilt',
        'GrLivArea',
        'FullBath',
        'BedroomAbvGr',
        'GarageCars',
        'GarageArea',
        'DistanceFromCity',
        'SalePrice'
    ]
]

y = data['Sold']
```

```python
sc = StandardScaler()

sc.fit(X)

scaled_data = sc.transform(X)

principal = PCA(n_components=3)

principal.fit(scaled_data)

x = principal.transform(scaled_data)

print(x.shape)
```

```python
principal.components_

plt.figure(figsize=(10,10))

plt.title('PCA 2D')

plt.scatter(
    x[:,0],
    x[:,1],
    c=data['Sold'],
    cmap='plasma'
)

plt.xlabel('PC1')
plt.ylabel('PC2')

plt.show()
```

```python
from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure(figsize=(10,10))

ax = fig.add_subplot(111, projection='3d')

ax.scatter(
    x[:,1],
    x[:,2],
    x[:,0],
    c=data['Sold'],
    cmap='plasma'
)

ax.set_xlabel('PC1')
ax.set_ylabel('PC2')
ax.set_zlabel('PC3')

plt.show()
```
